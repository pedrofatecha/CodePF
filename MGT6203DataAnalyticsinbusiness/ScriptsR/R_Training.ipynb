```r
#Creating Variables

a<- 1
b<- 2
c<- a + b

print(c)

exp(3)

log(exp(3))

```

```r
#logical

f1<- TRUE
f2<- FALSE
f3<- f1 & f2
f4<- f1 | f2
f5<- !f1

print(f3)
print(f4)
print(f5)
```

```r
#Vectors

myvec<- c(1, 2, 3, 4, 5)
myvec2<-1:10
myvec3<- seq(1, 10, by=2)
myvec4<- rep(c(1,2,3), times=5)

myvec_n<- c(myvec, myvec2, myvec3, myvec4)

for(i in 1:length(myvec_n)){
  print(myvec_n[i])
}
```
