Using July through October daily-high-temperature data for Atlanta for 1996 through 2015, use a CUSUM approach to identify when unofficial summer ends (i.e., when the weather starts cooling off) each year. >>2. Use a CUSUM approach to make a judgment of whether Atlanta’s summer climate has gotten warmer in that time (and if so, when).


```r
data_path_temp<-"/Users/pedrofatecha/GitHub/CodePF/ISYE6501_IntrotoAnalyticsModeling/01_Data/data 6.2/temps.txt"

Temperatures <- read.csv(data_path_temp,header = TRUE,sep = "\t")

#View(Temperatures)

Temperatures.columns<-colnames(Temperatures)[-1]
Temperature_transformed <- data.frame()   # start empty

for (i in names(Temperatures)[-1]) {      # loop over all cols except Date
  v_1 <- as.factor(Temperatures[[i]])     # temperatures
  v_2 <- as.factor(Temperatures[[1]])     # Date column
  year_val<-substr(i,start=2,stop=5)
  v_3 <- as.factor(rep(year_val, nrow(Temperatures)))  # column name repeated

  Temperature_transformed <- rbind(
    Temperature_transformed,
    data.frame(Date = v_2,
               year = v_3,
               temperature = v_1)
  )
}



Temperature_transformed[,"new_Date"]<-rep(NA, nrow(Temperature_transformed))


for(i in 1:nrow(Temperature_transformed)){
 Temperature_transformed[i,"new_Date"]<-paste(Temperature_transformed[i,1],Temperature_transformed[i,2],sep="-")
}

Temperature_transformed[,"new_Date"]<-as.Date(Temperature_transformed[,"new_Date"], format="%d-%b-%Y")




Temperature_transformed[,"temperature"]<-as.numeric(as.character(Temperature_transformed[,"temperature"]))

mu0<-mean(Temperature_transformed$temperature)
standard_dev<-sd(Temperature_transformed$temperature)


Temperature_transformed[,"deviation"]<-as.numeric(Temperature_transformed[,"temperature"] - mu0)
Temperature_transformed[,"deviation_squared"]<-as.numeric(Temperature_transformed[,"deviation"]/standard_dev)


#hist(Temperature_transformed$deviation_squared, main="Histogram for Temperature Deviation Squared", xlab="Deviation Squared", col="lightgreen", border="black")

Temperature_transformed[,"AlertUP"]<-rep("NA", nrow(Temperature_transformed))
Temperature_transformed[,"AlertDOWN"]<-rep("NA", nrow(Temperature_transformed))
Temperature_transformed[,"SumUP"]<-numeric()
Temperature_transformed[,"SumDOWN"]<-numeric()

T<-mean( Temperature_transformed$temperature) + 2.5*sd( Temperature_transformed$temperature)
k<-0
mu0<-mean( Temperature_transformed$temperature)

print(T)
print(mu0)

for(t in 1:nrow(Temperature_transformed)){
  S_plus  <- 0
  S_minus <- 0
  # recursion
    # upper (detect increase in mean)
    if(Temperature_transformed[t,"Date"]=="1-Jul"){
      S_plus <- max(0, Temperature_transformed[t,"temperature"] -mu0 - k)
      S_minus <- max(0, mu0 - k- Temperature_transformed[t,"temperature"]  )
      Temperature_transformed[t,"SumUP"]<-Temperature_transformed[t,"temperature"] -mu0 - k

      Temperature_transformed[t,"SumDOWN"]<-mu0 - k- Temperature_transformed[t,"temperature"]

    }
    else {
      S_plus  <- max(0,Temperature_transformed[t-1,"SumUP"] + (Temperature_transformed[t,"temperature"] -mu0 - k))
      # lower (detect decrease in mean)
      S_minus <- max(0,  Temperature_transformed[t-1,"SumDOWN"] + (mu0-k-Temperature_transformed[t,"temperature"]))

      Temperature_transformed[t,"SumUP"]<-Temperature_transformed[t-1,"SumUP"] + (Temperature_transformed[t,"temperature"] -mu0 - k)

      Temperature_transformed[t,"SumDOWN"]<-Temperature_transformed[t-1,"SumDOWN"] + (mu0-k-Temperature_transformed[t,"temperature"])
    }

    # check alarms
   if (S_plus > T ) {
      Temperature_transformed[t,"AlertUP"] <- TRUE
    } else if(S_minus < -T) {
      Temperature_transformed[t,"AlertDOWN"] <- FALSE
    }

}
years_Data<-c("1996",	"1997",	"1998",	"1999",	"2000",	"2001",	"2002",	"2003",	"2004",	"2005",	"2006",	"2007",	"2008",	"2009",	"2010",	"2011",	"2012",	"2013",	"2014",	"2015")

for(i in years_Data){
  data_temp<-Temperature_transformed[Temperature_transformed$year==i & Temperature_transformed$AlertUP==TRUE,]
  rownames(data_temp) <- NULL
  start_Date<-data_temp[1,"new_Date"]
  end_Date<-data_temp[nrow(data_temp),"new_Date"]
  print(paste("Year:",i," Start Date of Alarm:",start_Date," End Date of Alarm:",end_Date))
}

View(Temperature_transformed[Temperature_transformed$year=="1996" ,])

ggplot(Temperature_transformed, aes(x = new_Date, y = SumUP)) +
  geom_line(color = "blue") +

  geom_hline(yintercept = -T, linetype = "dashed", color = "red") +
  geom_hline(yintercept = T, linetype = "dashed", color = "red") +
  labs(
    title = "CUSUM Chart for Temperature Deviations (Upper)",
    x = "Date",
    y = "CUSUM Value"
  ) +
  scale_x_date(
    limits      = as.Date(c("2015-07-01", "2015-10-29")),  # your range
    date_breaks = "10 days",                               # or "1 day"
    date_labels = "%d-%b"                                  # e.g. "01-Jul"
  ) +
  theme_minimal()
```
