```r
library("kernlab")
library("kknn")
library("tidyr")
library("dplyr")
library("ggplot2")
library("plotly")
library("combinat")
library("car")
# Load Data Shared with the credit card applications
set.seed(1234)
data_path<- "/Users/pedrofatecha/GitHub/CodePF/MGT6203DataAnalyticsinbusiness/Data/UsedCars.csv"

Used_car <- read.table(data_path, sep = ",", header = TRUE)
```

```r
# Data Cleaning and Preprocessing

Model <- lm(Used_car$Price ~ Age + KM + HP + Metallic + Automatic + CC + Doors + Gears + Weight, data= Used_car)

Model.residuals <- resid(Model)
Model.fitted <- fitted(Model)
```

```r
Price_usedcar <- Used_car[1:10,"Price"]

Price_usedcar

Table_check<-cbind("Y"=Price_usedcar[1:10],"Y-hat"=as.vector(Model.fitted)[1:10],"Residuals"=as.vector(Model.residuals)[1:10])

Table_check<-as.data.frame(Table_check)

Table_check[,"Check"]<- Table_check[,"Y"] - Table_check[,"Y-hat"]

Table_check

```

```r
sum_detail<-summary(Model)

sum_detail
```

```r
beta_estimate <- sum_detail$coefficients[,1]
beta_estimate

Standard_error <- sum_detail$coefficients[,2]

tstat<- beta_estimate / Standard_error

cbind(tstat,sum_detail$coefficients[,3])

```

```r
# T statistic
Degrees_freedom <- Model$df.residual
Degrees_freedom
alpha<- 1- 0.95

1-(alpha/2)

qt(1-(alpha/2), Degrees_freedom)

qnorm(1-(alpha/2))

```

```r
pval<- 2 * pt(-abs(tstat), Degrees_freedom)

as.vector(pval)<=0.05

sig_ind<-ifelse(as.vector(pval)<=0.05, "significant","not significant")

cbind(Pvalue=pval,calc_Pvalue=sum_detail$coefficients[,4],tstat=sum_detail$coefficients[,3],significants=sig_ind)

sum_detail$coefficients[,1]/   sum_detail$coefficients[,2]
```

```r
# X point where we evaluate marginal effects
new_data <- data.frame(
  Income   = mean(Loan$Income),
  Family   = mean(Loan$Family),
  CCAvg    = mean(Loan$CCAvg),
  Education = factor(2, levels = c(1, 2, 3))
)

# Linear predictor Xβ (log-odds)
xb <- predict(logit_model, newdata = new_data, type = "link")

# Probability at that point
p_hat <- plogis(xb)   # same as predict(..., type="response")

p_hat
```

```r
lambda_prime <- dlogis(xb)  # derivative of logistic CDF at Xβ
lambda_prime
```

```r
beta_logit <- coef(logit_model)

# For continuous predictors, partial effect = dlogis(Xβ) * beta_j
pe_logit <- lambda_prime * beta_logit

pe_logit
```

```r
pe_logit_vars <- pe_logit[-1]   # drop intercept
pe_logit_vars
```
