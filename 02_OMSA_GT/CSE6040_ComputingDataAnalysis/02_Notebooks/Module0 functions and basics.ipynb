```python
import time
def mat_mul(A, B):
    """Multiply two 2x2 matrices."""
    return [
        [
            A[0][0] * B[0][0] + A[0][1] * B[1][0],
            A[0][0] * B[0][0] + A[0][1] * B[1][1],
        ],
        [
            A[1][0] * B[0][0] + A[1][1] * B[1][0],
            A[1][0] * B[0][0] + A[1][1] * B[1][1],
        ],
    ]

def mat_power(A, n):
    """Compute A^n using exponentiation by squaring."""
    result = [[1, 0], [0, 1]]  # Identity matrix

    while n > 0:
        if n % 2 == 1:
            result = mat_mul(result, A)

        A = mat_mul(A, A)
        n //= 2

    return result

def mat_vec_mul(A, v):
    """Multiply a 2x2 matrix by a 2x1 vector."""
    return [
        A[0][0] * v[0] + A[0][1] * v[1],
        A[1][0] * v[0] + A[1][1] * v[1],
    ]


def fibonacci(n):
    """
    Return the nth Fibonacci number.
    Defines F(0)=0, F(1)=1.
    """
    if n < 0:
        raise ValueError("n must be a non-negative integer")

    transition = [
        [1, 1],
        [1, 0],
    ]

    initial_state = [1, 0]  # [F(1), F(0)]

    power = mat_power(transition, n)
    state = mat_vec_mul(power, initial_state)

    return state[1]  # F(n)
mat_power([[1, 1], [1, 0]], 4)
```

```python
def fibonacci_naive(n):
    if n <= 1:
        return n
    return fibonacci_naive(n - 1) + fibonacci_naive(n - 2)


fibonacci_naive(10)
```
