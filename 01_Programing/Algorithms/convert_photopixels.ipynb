## Modify Pixels function

```python
from PIL import Image

# Open the image
img = Image.open('/Users/pedrofatecha/Downloads/generated-image.png')

# Define the new dimensions (width, height) in pixels
new_size = (720, 960)

# Resize the image
resized_img = img.resize(new_size, Image.Resampling.LANCZOS) # Use a high-quality resampling filter

# Save the new image (you can use a different filename to keep the original)
resized_img.save('/Users/pedrofatecha/Downloads/generated-image_fix.png')

print(f"Original size: {img.size}")
print(f"New size: {resized_img.size}")
```
